# Perceptron — Loss Function & Algorithm

## 1. What is a Perceptron?

A **Perceptron** is one of the simplest Machine Learning algorithms used for **binary classification**.

It takes input features, applies weights and a bias, and produces a prediction.

For two features:

$$
z = w_1x_1 + w_2x_2 + b
$$

Where:

* $x_1, x_2$ → input features
* $w_1, w_2$ → weights
* $b$ → bias
* $z$ → weighted sum

The Perceptron uses the sign of $z$ to determine the class.

---

# 2. Perceptron Classification

For the Perceptron algorithm, labels are represented as:

$$
y \in \{-1,+1\}
$$

The prediction is based on:

$$
z = w_1x_1 + w_2x_2 + b
$$

If:

$$
z > 0
$$

the model predicts:

$$
+1
$$

If:

$$
z < 0
$$

the model predicts:

$$
-1
$$

Therefore:

$$
\hat{y} =
\begin{cases}
+1 & \text{if } z > 0\\
-1 & \text{if } z < 0
\end{cases}
$$

---

# 3. Perceptron Loss Function

The Perceptron loss for one training example is:

$$
L = \max(0,-yz)
$$

where:

* $y$ → actual label (`-1` or `+1`)
* $z$ → weighted sum
* $L$ → loss

Since:

$$
z=w_1x_1+w_2x_2+b
$$

the loss can also be written as:

$$
L = \max(0,-y(w_1x_1+w_2x_2+b))
$$

---

# 4. Understanding the Perceptron Loss

The easiest way to understand the loss is to look at correct and incorrect predictions.

## Correct prediction

Suppose:

$$
y=+1
$$

and:

$$
z=+3
$$

Then:

$$
L=\max(0,-(+1)(+3))
$$

$$
L=\max(0,-3)
$$

Therefore:

$$
\boxed{L=0}
$$

The prediction is correct, so there is no loss.

---

## Incorrect prediction

Suppose:

$$
y=+1
$$

but:

$$
z=-3
$$

Then:

$$
L=\max(0,-(+1)(-3))
$$

$$
L=\max(0,3)
$$

Therefore:

$$
\boxed{L=3}
$$

The prediction is incorrect, so the loss is greater than zero.

---

# 5. Why Do We Check `z × y <= 0`?

In the Perceptron algorithm, we commonly see:

$$
zy \leq 0
$$

This condition identifies a misclassified example.

### Correct positive prediction

$$
y=+1,\quad z=+5
$$

Therefore:

$$
zy=(+1)(+5)=5
$$

Since:

$$
5>0
$$

the prediction is correct.

---

### Correct negative prediction

$$
y=-1,\quad z=-5
$$

Therefore:

$$
zy=(-1)(-5)=5
$$

Again:

$$
5>0
$$

The prediction is correct.

---

### Incorrect prediction

$$
y=+1,\quad z=-5
$$

Therefore:

$$
zy=(+1)(-5)=-5
$$

Since:

$$
-5\leq0
$$

the example is incorrectly classified.

Therefore, the model updates its parameters.

---

# 6. Relationship Between Loss and the Condition

The Perceptron loss is:

$$
L=\max(0,-yz)
$$

If:

$$
yz>0
$$

then:

$$
L=0
$$

The prediction is correct.

If:

$$
yz\leq0
$$

then:

$$
L\geq0
$$

and the example is considered misclassified by the Perceptron.

Therefore:

$$
\boxed{yz\leq0 \Rightarrow \text{Update the parameters}}
$$

This is why the implementation checks:

$$
z \times y \leq 0
$$

instead of explicitly calculating the loss every time.

---

# 7. Perceptron Weight Update

When the model makes a mistake, the weights are updated.

For two features:

$$
w_1=w_1+\eta yx_1
$$

$$
w_2=w_2+\eta yx_2
$$

The bias is updated as:

$$
b=b+\eta y
$$

where:

$$
\eta=\text{learning rate}
$$

The learning rate controls how large the update is.

---

# 8. Learning Rate

For example, if:

$$
\eta=0.1
$$

then the weight update becomes:

$$
w_1=w_1+0.1yx_1
$$

$$
w_2=w_2+0.1yx_2
$$

and:

$$
b=b+0.1y
$$

A larger learning rate produces larger updates.

A smaller learning rate produces smaller updates.

---

# 9. Perceptron Algorithm

The basic Perceptron algorithm is:

1. Initialize the weights and bias.
2. Select a training example.
3. Calculate the weighted sum.
4. Check whether the example is correctly classified.
5. If it is correct, do nothing.
6. If it is incorrect, update the weights and bias.
7. Repeat for the training examples.
8. Repeat for multiple iterations/epochs.
9. The final weights and bias define the decision boundary.

The overall flow is:

$$
X,y
\rightarrow
w,b
\rightarrow
z
\rightarrow
\text{Check classification}
\rightarrow
\text{Update if wrong}
\rightarrow
\text{Repeat}
$$

---

# 10. Weighted Sum

For two features:

$$
z=w_1x_1+w_2x_2+b
$$

For a particular observation:

$$
x_1=X[i][0]
$$

and:

$$
x_2=X[i][1]
$$

Therefore:

$$
z=w_1X[i][0]+w_2X[i][1]+b
$$

This value determines which side of the decision boundary the point lies on.

---

# 11. Decision Boundary

The Perceptron creates a **decision boundary** that separates the two classes.

The decision boundary occurs when:

$$
z=0
$$

Since:

$$
z=w_1x_1+w_2x_2+b
$$

the decision boundary is:

$$
w_1x_1+w_2x_2+b=0
$$

For two-dimensional data, this equation represents a **straight line**.

---

# 12. Converting the Decision Boundary to `y = mx + c`

Starting with:

$$
w_1x_1+w_2x_2+b=0
$$

Move the terms containing $x_1$ and $b$ to the other side:

$$
w_2x_2=-w_1x_1-b
$$

Divide by $w_2$:

$$
x_2=-\frac{w_1}{w_2}x_1-\frac{b}{w_2}
$$

The familiar equation of a straight line is:

$$
y=mx+c
$$

Therefore:

$$
\boxed{m=-\frac{w_1}{w_2}}
$$

and:

$$
\boxed{c=-\frac{b}{w_2}}
$$

So the Perceptron's decision boundary can be written as:

$$
\boxed{y=-\frac{w_1}{w_2}x-\frac{b}{w_2}}
$$

---

# 13. Example of the Decision Boundary

Suppose the trained Perceptron gives:

$$
w_1=1.0066
$$

$$
w_2=0.2222
$$

$$
b=1.3
$$

Then:

$$
m=-\frac{1.0066}{0.2222}
$$

$$
m\approx-4.5313
$$

And:

$$
c=-\frac{1.3}{0.2222}
$$

$$
c\approx-5.8519
$$

Therefore:

$$
\boxed{y=-4.5313x-5.8519}
$$

This is the line plotted on the scatter plot.

---

# 15. Complete Conceptual Flow

The complete Perceptron process can be remembered as:

$$
\boxed{
\text{Input Data}
\rightarrow
\text{Weighted Sum}
\rightarrow
\text{Prediction}
\rightarrow
\text{Loss/Error}
\rightarrow
\text{Weight Update}

}
$$

After training:

$$
\boxed{
w_1,w_2,b
\rightarrow
\text{Decision Boundary}
\rightarrow
\text{Visualization}
}
$$

# 18. Key Takeaway

The **Perceptron** learns a decision boundary by repeatedly correcting misclassified training examples.

The core idea is:

$$
\boxed{
\text{Prediction}
\rightarrow
\text{Check Error}
\rightarrow
\text{Update Weights}
\rightarrow
\text{Repeat}
}
$$

The final weights and bias define the decision boundary:

$$
\boxed{w_1x_1+w_2x_2+b=0}
$$

For two-dimensional data, this becomes a straight line:

$$
\boxed{y=-\frac{w_1}{w_2}x-\frac{b}{w_2}}
$$

So the **scatter plot shows the actual data**, while the **red line shows the decision boundary learned by the Perceptron**.


```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import os
import kagglehub
import matplotlib.pyplot as plt 
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,
                          n_redundant=0, n_classes=2, n_clusters_per_class=1,
                          random_state=41, hypercube=False, class_sep=15)
```

```python
X.shape
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="winter", s=50)
```

```python
def Perceptron(X, y):

    w1=w2=b=1
    lr = 0.1

    for j in range(10000):
        for i in range(X.shape[0]):
            z = w1*X[i][0] + w2*X[i][1] + b

            if z*y[i] <= 0:
                w1 = w1 + lr*y[i]*X[i][0]
                w2 = w2 + lr*y[i]*X[i][1]
                b = b + lr*y[i]


    return w1, w2, b
```

```python
w1, w2, b = Perceptron(X, y)
m = -(w1/w2)
c = -(b/w2)
print(f"Value of w1: {w1} \nValue of w2: {w2} \nValue of b: {b}")
print(f"Value of m is: {m} \nValue of c: {c}")
```

```python
x_input = np.linspace(-3,3,100)
y_input = (m*x_input) + c
plt.figure(figsize=(10,6))
plt.plot(x_input, y_input, color="red")
plt.scatter(X[:,0], X[:, 1],c=y, cmap="winter")
plt.ylim(-3,2)
```
